```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ed= pd.read_csv("exam.csv")

```

```python
new_person = pd.DataFrame({"study_hours": [6.5], "past_mark": [60]})

colours = ed["passed"].map({0: "red", 1: "seagreen"})
plt.scatter(ed["study_hours"], ed["passed"], c=colours, s=200, edgecolors="black")
for name, row in ed.iterrows():
    plt.annotate(name, (row["study_hours"], row["passed"]), xytext=(0, 15), textcoords="offset points", ha="center")
plt.scatter(6.5, 1, marker="*", s=600, color="gold", edgecolors="black", label="NEW person")
plt.xlabel("Study Hours")
plt.ylabel("Passed")
plt.title("8 applicants (green = passed, red = failed)")
plt.legend()
plt.show()
```
